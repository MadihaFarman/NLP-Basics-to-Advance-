### What we are actually going to do
Three things in order:

Download GloVe vectors that were pretrained on Wikipedia + Gigaword (6 billion words) — the 50-dimensional version. These are vectors that someone else already trained, so you get the benefit of massive training data without training anything yourself.


Run similarity queries — give it two words and get back a number between -1 and 1 telling you how similar they are in vector space. You will run pairs that should be similar and pairs that should be distant and verify the scores make intuitive sense.


Run arithmetic — the famous king - man + woman = queen example. You give it positive words and negative words, and it finds the word whose vector is closest to the result of that arithmetic.

## Loading Pre-trained model

```python
from gensim.downloader import load
model = load("glove-wiki-gigaword-50") 
```

### An Insight I got

As GloVe calculates vector similarity based on word co-occurrence, I discovered a highly problematic association: the sacred word 'Allah' shares a 70% similarity score with a severe profanity (Na'uzubillah). This reflects how malicious data and deliberate disrespect in the training corpus introduces severe biases into embeddings. Consequently, if this model were deployed in a next-word prediction system, it could catastrophically suggest highly offensive terms in the top recommendations.
