## Text Preprocessing :Tokenization , Stemming , Lemmatization 

### Preprocessing pipeline: raw text → tokens → stems or lemmas → model

```python
import re

def tokenize(text):
    return re.findall(r"[A-Za-z]+(?:'[A-Za-z]+)?|[0-9]+|[^\sA-Za-z0-9]", text)
```

```python
tokenize("The cats weren't running at 3pm.")
```

```python
def stem_step_1a(word):
    if word.endswith("sses"):
        return word[:-2]
    if word.endswith("ies"):
        return word[:-2]
    if word.endswith("ss"):
        return word
    if word.endswith("s") and len(word) > 1:
        return word[:-1]
    return word
```

```python
def stem_step_1b(word):
    if any(char in "aeiouAEIOU" for char in word) and word[-2:]=="ed":
        return word[:-2]
    if any(char in "aeiouAEIOU" for char in word) and word[-3:]=="ing":
        nword = word[:-3]
        lastCons = nword[-1]
        secondLastCons = nword[-2]
        if lastCons == secondLastCons:
            return nword[:-1]
        else :
            return nword


print(stem_step_1b("educated"))
        
```

```python
LEMMA_TABLE = {
    ("running", "VERB"): "run",
    ("ran", "VERB"): "run",
    ("runs", "VERB"): "run",
    ("better", "ADJ"): "good",
    ("best", "ADJ"): "good",
    ("cats", "NOUN"): "cat",
    ("cat", "NOUN"): "cat",
    ("were", "VERB"): "be",
    ("was", "VERB"): "be",
    ("is", "VERB"): "be",
}

def lemmatize(word, pos):
    key = (word.lower(), pos)
    if key in LEMMA_TABLE:
        return LEMMA_TABLE[key]
    if pos == "VERB" and word.endswith("ing"):
        return word[:-3]
    if pos == "NOUN" and word.endswith("s"):
        return word[:-1]
    return word.lower()
```

```python
lemmatize("running", "VERB")
```

```python
lemmatize("eating","VERB")
```

```python
def preprocess(text, pos_tagger=None):
    tokens = tokenize(text)
    stems = [stem_step_1a(t.lower()) for t in tokens]
    tags = pos_tagger(tokens) if pos_tagger else [(t, "NOUN") for t in tokens]
    lemmas = [lemmatize(word, pos) for word, pos in tags]
    return {"tokens": tokens, "stems": stems, "lemmas": lemmas}
```

```python
preprocess("Madiha wanted to running after cats")
```

WordNet® is a large lexical database of English. Nouns, verbs, adjectives and adverbs are grouped into sets of cognitive synonyms (synsets), each expressing a distinct concept. Synsets are interlinked by means of conceptual-semantic and lexical relations. 

spaCy'smorphologizer isa trainable spaCy pipeline component that predicts morphological features and coarse-grained Part-of-Speech (POS) tags for tokens in a document

```python
import nltk
nltk.download("punkt_tab")
nltk.download("wordnet")
nltk.download("averaged_perceptron_tagger_eng")

from nltk.tokenize import word_tokenize
from nltk.stem import PorterStemmer, WordNetLemmatizer
from nltk import pos_tag

text = "The cats were running."
tokens = word_tokenize(text)
stems = [PorterStemmer().stem(t) for t in tokens]
lemmatizer = WordNetLemmatizer()
tagged = pos_tag(tokens)


def nltk_pos_to_wordnet(tag):
    if tag.startswith("V"):
        return "v"
    if tag.startswith("J"):
        return "a"
    if tag.startswith("R"):
        return "r"
    return "n"


lemmas = [lemmatizer.lemmatize(t, nltk_pos_to_wordnet(tag)) for t, tag in tagged]
```

## Normalization

```python
import re
import unicodedata
def normalise(text):
 text = text.lower()
 text = re.sub(r'http\S+', '', text)
 text = re.sub(r'<.*?>', '', text)
 text = re.sub(r'[^a-z0-9\s]', '', text)
 text = re.sub(r'\s+', ' ', text).strip()
 return text
print(normalise("Check out <b>NLP</b>!! Visit https://huggingface.co"))
```
