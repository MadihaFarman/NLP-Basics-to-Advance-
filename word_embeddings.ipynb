## Word Embeddings

### Word2Vec

```python
from gensim.models import Word2Vec
text = "Dr. Madiha's research on NLP is state-of-the-art. She's published 2+ papers!"
sentences = [['nlp', 'is', 'capturing', 'meaning'], ['machine', 'learning', 'systems']]
model = Word2Vec(sentences, vector_size=100, window=5, min_count=1, sg=1, epochs=10)
print("Similarity score:", model.wv.similarity('machine', 'learning'))
```

```python
print("Most similar words to 'nlp':", model.wv.most_similar('nlp'))
```

```python
from sklearn.datasets import fetch_20newsgroups
from sklearn.feature_extraction.text import CountVectorizer,TfidfVectorizer
newsgroups = fetch_20newsgroups()
# viewing the overall structure of the dataset
# newsgroups.keys()
# newsgroups.data[0]  # viewing the first document in the dataset
vectorizer1 = CountVectorizer()
vectorizer2 = TfidfVectorizer()
X_count = vectorizer1.fit_transform(newsgroups.data)
X_tfidf = vectorizer2.fit_transform(newsgroups.data)
print("Shape of Count Vectorizer output:", X_count.shape)
print("Shape of TF-IDF Vectorizer output:", X_tfidf.shape)
# print("Feature names from Count Vectorizer:", vectorizer1.get_feature_names_out())
# print("Feature names from TF-IDF Vectorizer:", vectorizer2.get_feature_names_out())  
print("Count Vectorizer output for the first document:\n", X_count[0].toarray())
print("TF-IDF Vectorizer output for the first document:\n", X_tfidf[0].toarray())   
```
